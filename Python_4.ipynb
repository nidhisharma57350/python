```python
#Fabonacci Series:
#In fabonacci series first two terms are fixed ,which are 0 and 1.Then we find next term from these two terms by adding them .In next step
#the term which was second term in first step become the first term and the currently found third term become the second term.Continue this process
#to get required number of terms.
```

```python
def fabonacci(n):
   first_term=0
   second_term=1
   print(first_term)
   print(second_term)
   for i in range(3,n+1):
      third_term=first_term+second_term
      print(third_term)
      first_term=second_term
      second_term=third_term
fabonacci(10)
```
